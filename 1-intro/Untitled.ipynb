```python
median_before = df["horsepower"].median()
most_frequent = df["horsepower"].mode().iloc[0]
horsepower_filled = df["horsepower"].fillna(most_frequent)
median_after = horsepower_filled.median()
print("Медиана до:", median_before)
print("Самое частое значение:", most_frequent)
print("Медиана после:", median_after)
if median_after > median_before:
    print("Yes, it increased")
elif median_after < median_before:
    print("Yes, it decreased")
else:
    print("No")
```

```python
X = (
    df.loc[df["origin"] == "Asia", ["vehicle_weight", "model_year"]]
    .head(7)
    .to_numpy(dtype=float)
)

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y

print("Веса:", w)
print("Сумма весов:", w.sum())
print("С округлением:", round(w.sum(), 4))
```
